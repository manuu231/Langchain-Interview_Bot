```python
# All imports
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage, trim_messages
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.output_parsers import StrOutputParser
from IPython.display import display, Markdown
```

```python
llm = ChatGoogleGenerativeAI(
     model = "gemini-2.5-flash",
     temperature = 0.3
 )
prompt = ChatPromptTemplate.from_messages([
    ("system","You are a technical interviewer for data science roles. Ask one question at a time. Give feedback after each answer."),
    MessagesPlaceholder(variable_name = "history"),
    ("human","{input}")
])

parser = StrOutputParser()
chain = prompt|llm | parser

store = {}

def get_session_history(session_id: str) -> ChatMessageHistory:
  if session_id not in store:
    store[session_id] = ChatMessageHistory()
  return store[session_id]

main_chain = RunnableWithMessageHistory(
     chain,
     get_session_history,
     input_messages_key = "input",
     history_messages_key = "history"
 )
```

```python
# Round 1: Start interview
response = main_chain.invoke(
    {"input": "Hi, I am ready. Please start the interview."},
    config={"configurable": {"session_id": "interview_session_1"}}
)
print("INTERVIEWER:", response)

# Round 2: Sample answer to first question
response = main_chain.invoke(
    {
        "input": """
        Python is my primary programming language. I have used it extensively
        for data analysis, machine learning, and deep learning projects.
        I am comfortable with libraries such as Pandas, NumPy, Scikit-learn,
        Matplotlib, and PyTorch.
        """
    },
    config={"configurable": {"session_id": "interview_session_1"}}
)
print("\nINTERVIEWER:", response)

# Round 3: Sample answer to second question
response = main_chain.invoke(
    {
        "input": """
        One machine learning project I worked on was a biometric authentication
        system using PPG signals. I built a hybrid deep learning model combining
        CNN and LSTM architectures to extract spatial and temporal features.
        The model achieved strong performance with ROC-AUC close to 0.99 on the
        CFISHR dataset.
        """
    },
    config={"configurable": {"session_id": "interview_session_1"}}
)
print("\nINTERVIEWER:", response)
```
